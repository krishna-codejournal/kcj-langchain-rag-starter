````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
# 2D example of embeddings
word_embeddings = {
    "king": np.array([0.8, 0.6]),
    "queen": np.array([0.7, 0.5]),
    "man": np.array([0.9, 0.4]),
    "woman": np.array([0.8, 0.3]),
    "apple": np.array([0.1, 0.9]),
    "banana": np.array([0.2, 0.8])
}
````

````python
fig, ax = plt.subplots(figsize=(8, 6))

for word, vec in word_embeddings.items():
    ax.scatter(vec[0], vec[1],s=100)
    ax.annotate(word, (vec[0],vec[1]), xytext=(5,5), textcoords='offset points') 
ax.set_xlabel('Dimension 1')
ax.set_ylabel('Dimension 2')
ax.set_title('2D Word Embeddings Visualization')
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.show()
````

# 🧩 Understanding a 2D Embedding Example

Below is a **toy example** used to explain how embeddings work conceptually.

```python
# 2D example of embeddings
word_embeddings = {
    "king": np.array([0.8, 0.6]),
    "queen": np.array([0.7, 0.5]),
    "man": np.array([0.9, 0.4]),
    "woman": np.array([0.8, 0.3]),
    "apple": np.array([0.1, 0.9]),
    "banana": np.array([0.2, 0.8])
}
```

---

## 1️⃣ What This Example Represents

* Each **word** is represented as a **2‑dimensional vector**
* Each vector position is a **learned semantic feature**
* This is a **simplified visualization** of real embedding models

⚠️ Real-world embedding models typically use **hundreds or thousands of dimensions**.

---

## 2️⃣ Interpreting the Dimensions (Intuition Only)

The dimensions are **not explicitly labeled**, but for understanding we can loosely imagine:

* **X‑axis** → Human / Entity related
* **Y‑axis** → Nature / Abstract related

> Important: In real models, dimensions have **no human-readable meaning**. Meaning emerges from training data.

---

## 3️⃣ Semantic Grouping in the Vector Space

### 👑 Human & Royal Concepts

```
king   → [0.8, 0.6]
queen  → [0.7, 0.5]
man    → [0.9, 0.4]
woman  → [0.8, 0.3]
```

* High X values
* Close proximity to each other
* Represent related meanings

### 🍎 Fruits

```
apple  → [0.1, 0.9]
banana → [0.2, 0.8]
```

* Low X, high Y values
* Clustered together
* Far from human-related words

---

## 4️⃣ Visual Intuition (2D Space)

```
Y ↑
1.0 |        apple ●
0.9 |     banana ●
0.8 |
0.7 |           king ●
0.6 |        queen ●
0.5 |
0.4 |      man ●
0.3 |   woman ●
    +----------------------→ X
      0.1   0.5   0.9
```

* Similar words form **clusters**
* Distant words have **different meanings**

---

## 5️⃣ Similarity Explained

* **king ↔ queen** → high cosine similarity (same direction)
* **king ↔ apple** → low similarity (different direction)

This is how semantic search, recommendations, and RAG systems work internally.

---

## 6️⃣ Relationships in Embeddings

In real embedding spaces, we can capture relationships such as:

```
king − man + woman ≈ queen
```

This is possible because embeddings encode **relationships**, not just similarity.

---

## 7️⃣ Why Fruits Are Far Away

During training:

* "apple" appears near words like *fruit, food, sweet*
* "king" appears near *royal, crown, palace*

As a result, embeddings naturally separate them into **different regions of space**.

---

## 8️⃣ What This Example Teaches

| Concept                     | Demonstrated |
| --------------------------- | ------------ |
| Words → vectors             | ✅            |
| Similar meanings cluster    | ✅            |
| Different meanings separate | ✅            |
| Distance represents meaning | ✅            |

---

## 9️⃣ Key Takeaway

> **Embeddings convert meaning into geometry, allowing machines to compare semantics using distance and direction.**
