# 🧮 Cosine Similarity Explained with a 2D Embedding Example

This document explains **how cosine similarity is calculated** using a simple **2D word embedding example**.

---

## 1️⃣ Given 2D Word Embeddings

```python
king   = [0.8, 0.6]
queen  = [0.7, 0.5]
man    = [0.9, 0.4]
woman  = [0.8, 0.3]
apple  = [0.1, 0.9]
banana = [0.2, 0.8]
```

Each word is represented as a **2D vector**. The goal is to measure **semantic similarity** by comparing vector directions.

---

## 2️⃣ Cosine Similarity Formula

Cosine similarity between two vectors **A** and **B** is defined as:

```
cosine_similarity = (A · B) / (|A| × |B|)
```

Where:

* `A · B` is the **dot product**
* `|A|` and `|B|` are the **magnitudes (lengths)** of the vectors

---

## 3️⃣ Example 1: king vs queen (High Similarity)

### Step 1: Dot product

```
(0.8 × 0.7) + (0.6 × 0.5) = 0.56 + 0.30 = 0.86
```

### Step 2: Magnitudes

```
|king|  = √(0.8² + 0.6²) = 1.0
|queen| = √(0.7² + 0.5²) ≈ 0.86
```

### Step 3: Cosine similarity

```
0.86 / (1.0 × 0.86) = 1.0
```

✅ *king* and *queen* are semantically very close.

---

## 4️⃣ Example 2: king vs man (High but Slightly Lower)

### Dot product

```
(0.8 × 0.9) + (0.6 × 0.4) = 0.72 + 0.24 = 0.96
```

### Magnitudes

```
|man| = √(0.9² + 0.4²) ≈ 0.98
```

### Cosine similarity

```
0.96 / (1.0 × 0.98) ≈ 0.98
```

✅ Strong similarity, but *king* includes an extra concept (royalty).

---

## 5️⃣ Example 3: king vs apple (Low Similarity)

### Dot product

```
(0.8 × 0.1) + (0.6 × 0.9) = 0.08 + 0.54 = 0.62
```

### Magnitudes

```
|apple| = √(0.1² + 0.9²) ≈ 0.91
```

### Cosine similarity

```
0.62 / (1.0 × 0.91) ≈ 0.68
```

🔴 Much lower similarity → different semantic category.

---

## 6️⃣ Comparison Summary

| Word Pair    | Cosine Similarity | Interpretation           |
| ------------ | ----------------- | ------------------------ |
| king ↔ queen | ~1.00             | Almost identical meaning |
| king ↔ man   | ~0.98             | Strongly related         |
| king ↔ apple | ~0.68             | Weakly related           |

---

## 7️⃣ Why Cosine Similarity Works for Embeddings

* Compares **direction**, not magnitude
* Handles different vector lengths
* Works the same in **2D** and **3000D** spaces

This makes it ideal for:

* Semantic search
* Recommendation systems
* Clustering
* Retrieval-Augmented Generation (RAG)

---

## 8️⃣ Key Takeaway

> **Cosine similarity measures semantic similarity by checking how closely two embedding vectors point in the same direction.**

---

✅ This document provides a clear, step-by-step explanation suitable for learning notes, interviews, and AI fundamentals.


````python
import numpy as np

def cosine_similarity(vec1, vec2):
    dot_product = np.dot(vec1, vec2)
    norm_vec1 = np.linalg.norm(vec1)
    norm_vec2 = np.linalg.norm(vec2)
    return dot_product / (norm_vec1 * norm_vec2)
````

````python
# 2D example of embeddings
word_embeddings = {
    "king": np.array([0.8, 0.6]),
    "queen": np.array([0.7, 0.5]),
    "man": np.array([0.9, 0.4]),
    "woman": np.array([0.8, 0.3]),
    "apple": np.array([0.1, 0.9]),
    "banana": np.array([0.2, 0.8])
}
````

````python
king_queen_similarity = cosine_similarity(word_embeddings["king"], word_embeddings["queen"])
print(f"Cosine similarity between 'king' and 'queen': {king_queen_similarity:.4f}")

king_man_similarity = cosine_similarity(word_embeddings["king"], word_embeddings["man"])
print(f"Cosine similarity between 'king' and 'man': {king_man_similarity:.4f}") 

king_woman_similarity = cosine_similarity(word_embeddings["king"], word_embeddings["woman"])
print(f"Cosine similarity between 'king' and ' woman': {king_woman_similarity:.4f}")

king_banana_similarity = cosine_similarity(word_embeddings["king"], word_embeddings["banana"])
print(f"Cosine similarity between 'king' and 'banana': {king_banana_similarity:.4f}")   

banana_woman_similarity = cosine_similarity(word_embeddings["banana"], word_embeddings["woman"])
print(f"Cosine similarity between 'banana' and 'woman': {banana_woman_similarity:.4f}")

apple_banana_similarity = cosine_similarity(word_embeddings["apple"], word_embeddings["banana"])
print(f"Cosine similarity between 'apple' and 'banana': {apple_banana_similarity:.4f}")
````
