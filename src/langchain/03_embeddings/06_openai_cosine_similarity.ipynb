```python
import numpy as np

def cosine_similarity(vec1, vec2):
    """Compute the cosine similarity between two vectors."""
    dot_product = np.dot(vec1, vec2)
    norm_vec1 = np.linalg.norm(vec1)
    norm_vec2 = np.linalg.norm(vec2)
    if norm_vec1 == 0 or norm_vec2 == 0:
        return 0.0
    return dot_product / (norm_vec1 * norm_vec2)
```

```python
sentences = ["LangChain is a powerful framework for building applications with LLMs.",
             "AI applications uses large language models to perform various tasks.",
             "LangChain simplifies processing and managing language models.",
             "Embeddings are a way to represent text in a numerical format.",
             "Cosine similarity measures the similarity between two vectors.",
             "This weekend I plan to go hiking in the mountains.",
             "The best shoes for hiking are waterproof and have good ankle support."]
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small" 
)

sentence_embeddings = embeddings.embed_documents(sentences)
```

```python
# Calculate cosine similarities between all sentence embeddings 
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        sim = cosine_similarity(sentence_embeddings[i], sentence_embeddings[j])
        print(f"Cosine Similarity between Sentence {i+1} and Sentence {j+1}: {sim:.4f}")
```

```python
def semantic_search(query, documents,embedding_models, top_k=3):
    """Perform semantic search to find the most relevant documents to the query."""
    query_embedding = embedding_models.embed_query(query)
    document_embeddings = embedding_models.embed_documents(documents)

    similarities = []
    for i, doc_embedding in enumerate(document_embeddings):
        similarity = cosine_similarity(query_embedding, doc_embedding)
        similarities.append((similarity, documents[i]))

    similarities.sort(reverse=True, key=lambda x: x[0])
    return similarities[:top_k]    
```

```python
query = "What framework helps build applications with LLMs?"
print("\nSemantic Search Results: for Query ->", query)
results = semantic_search(query, sentences, embeddings)
for similarity, document in results:
    print(f"Similarity: {similarity:.4f}, Document: {document}")
```
